# NumPy & Pandas Deep Dive

## Objective
The objective of this notebook is to practice NumPy and Pandas operations including array creation, statistics, reshaping, slicing, broadcasting, data loading, data cleaning, and data merging.

```python
import numpy as np
import pandas as pd

print("Libraries Imported Successfully")
```

# NumPy Practice
## Array Creation

```python
numbers = np.random.randint(1, 101, 20)

print("Random Numbers:")
print(numbers)
```

## Statistics Calculation

```python
print("Mean:", np.mean(numbers))
print("Median:", np.median(numbers))
print("Standard Deviation:", np.std(numbers))
print("Minimum:", np.min(numbers))
print("Maximum:", np.max(numbers))
```

## Reshaping Arrays

```python
matrix = numbers.reshape(4, 5)

print(matrix)
```

## Array Slicing

```python
print(matrix[:2])
```

## Broadcasting

```python
broadcasted = matrix * 2

print(broadcasted)
```

# Pandas Practice
## Data Loading

```python
df = pd.read_csv("../data/titanic.csv")

df.head()
```

# Dataset Inspection

In this section, we inspect the structure of the dataset including shape, columns, data types, and statistical information.

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

```python
df.describe()
```

# Missing Values Analysis

Checking missing values present in the dataset.

```python
df.isnull().sum()
```

# Data Cleaning

Handling missing values and removing duplicate records.

```python
df["Age"] = df["Age"].fillna(
    df["Age"].median()
)

df["Embarked"] = df["Embarked"].fillna(
    "Unknown"
)

df["Cabin"] = df["Cabin"].fillna(
    "Not Available"
)

df = df.drop_duplicates()

print(
    "Cleaning Completed Successfully"
)
```

```python
df.isnull().sum()
```

# Data Type Verification

Verifying data types after cleaning.

```python
df.dtypes
```

# Data Merging

Merging Titanic dataset with department information dataset.

```python
department_df = pd.read_csv(
    "../data/department_info.csv"
)

department_df.head()
```

```python
merged_df = pd.merge(
    df,
    department_df,
    on="Pclass",
    how="left"
)

merged_df.head()
```

# Merged Dataset Information

Checking shape and columns after merging.

```python
print("Merged Shape:")
print(merged_df.shape)

print("\nColumns:")
print(merged_df.columns)
```

# Dataset Summary

Generating complete statistical summary of the dataset.

```python
merged_df.describe(include="all")
```
