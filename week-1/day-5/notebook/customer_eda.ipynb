# 4. Data Cleaning

Data cleaning improves the quality of analysis by handling missing values and removing duplicate records.

```python
df["Income"] = df["Income"].fillna(df["Income"].median())
```

```python
df = df.drop_duplicates()
```

```python
df["Age"] = 2026 - df["Year_Birth"]
```

```python
df.isnull().sum()
```

## Cleaning Decisions

- Missing values in the Income column were replaced using the median.
- Duplicate rows were removed.
- A new Age feature was created from the Year_Birth column.

# 6. Data Visualization

Visualizations help identify trends, patterns, and relationships within the dataset.

## Visualization 1: Age Distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"], bins=20, kde=True)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

### Interpretation

Most customers belong to the middle-age group, indicating that the company mainly targets mature customers.

## Visualization 2: Education Level Distribution

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Education")

plt.title("Distribution of Education Levels")
plt.xticks(rotation=45)

plt.show()
```

### Interpretation

Graduation and PhD are among the most common education levels in the dataset.

## Visualization 3: Income Distribution by Education

```python
plt.figure(figsize=(10,5))

sns.boxplot(data=df, x="Education", y="Income")

plt.xticks(rotation=45)

plt.title("Income Distribution by Education")

plt.show()
```

### Interpretation

Customers with higher educational qualifications generally have higher incomes.

## Visualization 4: Correlation Heatmap

```python
plt.figure(figsize=(12,8))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Matrix")

plt.show()
```

### Interpretation

Several spending variables show positive relationships with customer income.

## Visualization 5: Income vs Wine Spending

```python
plt.figure(figsize=(8,5))

sns.regplot(data=df,
            x="Income",
            y="MntWines")

plt.title("Income vs Wine Spending")

plt.show()
```

## Visualization 6: Income Distribution by Marital Status

```python
plt.figure(figsize=(10,5))

sns.violinplot(
    data=df,
    x="Marital_Status",
    y="Income"
)

plt.xticks(rotation=45)
plt.title("Income Distribution by Marital Status")

plt.show()
```

### Interpretation

Customers with higher incomes tend to spend more on wine products, indicating a positive relationship between income and spending.
