# Data Visualization using Matplotlib and Seaborn

## Objective

The objective of this notebook is to explore and visualize the Titanic dataset using various visualization techniques. Different chart types are used to understand distributions, relationships, patterns, and correlations within the dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries Imported Successfully")
```

# Dataset Loading

Loading the Titanic dataset for visualization and analysis.

```python
df = pd.read_csv("../data/titanic.csv")

df.head()
```

# Dataset Information

Understanding dataset dimensions and data types.

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

# Histogram

Visualizing the distribution of passenger ages.

```python
plt.figure(figsize=(8,5))

plt.hist(
    df["Age"].dropna(),
    bins=15
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

# Bar Chart

Showing passenger count by class.

```python
df["Pclass"].value_counts().plot(
    kind="bar"
)

plt.title(
    "Passenger Class Distribution"
)

plt.xlabel(
    "Passenger Class"
)

plt.ylabel(
    "Count"
)

plt.show()
```

# Box Plot

Analyzing Age and Fare distributions.

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df[["Age","Fare"]]
)

plt.title(
    "Age and Fare Distribution"
)

plt.show()
```

# Correlation Heatmap

Visualizing relationships among numerical variables.

```python
numeric_df = df.select_dtypes(
    include="number"
)

plt.figure(figsize=(10,6))

sns.heatmap(
    numeric_df.corr(),
    annot=True
)

plt.title(
    "Correlation Heatmap"
)

plt.show()
```

# Scatter Plot

Analyzing the relationship between Age and Fare.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare"
)

plt.title(
    "Age vs Fare"
)

plt.show()
```

# Violin Plot

Understanding fare distribution across passenger classes.

```python
plt.figure(figsize=(8,5))

sns.violinplot(
    data=df,
    x="Pclass",
    y="Fare"
)

plt.title(
    "Fare Distribution by Passenger Class"
)

plt.show()
```

# Key Insights

### Observations

- Most passengers were between 20 and 40 years old.
- Passenger Class 3 contained the highest number of travelers.
- Fare values contain several outliers.
- Passenger class significantly affects fare distribution.
- Numerical features show varying levels of correlation.
- Data visualization helps identify trends and patterns effectively.

# Conclusion

This notebook demonstrated the use of Matplotlib and Seaborn for data visualization. Multiple chart types including Histogram, Bar Chart, Box Plot, Heatmap, Scatter Plot, and Violin Plot were used to analyze the Titanic dataset and extract meaningful insights.

These visualization techniques are essential for Exploratory Data Analysis (EDA) and Machine Learning workflows.
