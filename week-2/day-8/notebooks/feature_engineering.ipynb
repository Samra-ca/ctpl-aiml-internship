# Feature Engineering & Data Preprocessing

## Objective

The objective of this task is to prepare the dataset for machine learning by applying preprocessing and feature engineering techniques.

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('../data/WA_Fn-UseC_-HR-Employee-Attrition.csv')

df.head()
```

# Dataset Overview

This section provides basic information about the dataset including shape, columns, and data types.

```python
print("Shape:", df.shape)

df.info()
```

# Missing Value Analysis

Checking whether the dataset contains missing values before preprocessing.

```python
df.isnull().sum()
```

# Feature Engineering

Creating a new feature named ExperienceRatio using TotalWorkingYears and Age.

```python
df["ExperienceRatio"] = (
    df["TotalWorkingYears"] /
    df["Age"]
)

df[["Age", "TotalWorkingYears", "ExperienceRatio"]].head()
```

# Encoding Categorical Variables

Machine learning models require numerical input. Therefore, categorical columns are encoded.

```python
categorical_cols = df.select_dtypes(
    include="object"
).columns

categorical_cols
```

```python
le = LabelEncoder()

for col in categorical_cols:
    df[col] = le.fit_transform(df[col])

df.head()
```

# Feature Scaling

Standardizing numerical features using StandardScaler.

```python
scaler = StandardScaler()

numerical_cols = df.select_dtypes(
    include=np.number
).columns

df[numerical_cols] = scaler.fit_transform(
    df[numerical_cols]
)

df.head()
```

# Dataset After Preprocessing

The dataset is now ready for machine learning model development.

```python
df.describe()
```
