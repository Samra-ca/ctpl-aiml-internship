# Statistical Analysis & Hypothesis Testing

## Objective

The objective of this task is to perform statistical analysis and hypothesis testing on employee data to uncover meaningful insights.

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats

df = pd.read_csv('../data/WA_Fn-UseC_-HR-Employee-Attrition.csv')

df.head()
```

# Descriptive Statistics

Calculating mean, median, mode, and standard deviation for Monthly Income.

```python
print("Mean:", df["MonthlyIncome"].mean())
print("Median:", df["MonthlyIncome"].median())
print("Mode:", df["MonthlyIncome"].mode()[0])
print("Standard Deviation:", df["MonthlyIncome"].std())
```

# Distribution Analysis

Visualizing Monthly Income distribution.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["MonthlyIncome"],
    kde=True
)

plt.title("Monthly Income Distribution")

plt.show()
```

# Correlation Analysis

Analyzing relationships among numerical variables.

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

# Hypothesis Testing

Comparing Monthly Income of employees who stayed versus employees who left.

```python
yes_attrition = df[df["Attrition"] == "Yes"]["MonthlyIncome"]

no_attrition = df[df["Attrition"] == "No"]["MonthlyIncome"]

t_stat, p_value = stats.ttest_ind(
    yes_attrition,
    no_attrition
)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

# Interpretation

If p-value < 0.05:

- Reject Null Hypothesis
- Significant difference exists

If p-value > 0.05:

- Fail to Reject Null Hypothesis
- No significant difference exists
